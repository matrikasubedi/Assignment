CASE STUDY 2: \
1: Top Product per Category\
You have a table name is sales with column namecategory, product and revenue\
Table structure:\
('Electronics', 'Phone', 1000),\
('Electronics', 'Laptop', 2000),\
('Electronics', 'Tablet', 1500),\
('Clothing', 'Shirt', 500),\
('Clothing', 'Jacket', 800)\
Problem – Find highest selling product in each category.


```python
%sql
-- Create table sales
CREATE OR REPLACE TABLE sales (
    category STRING,
    product STRING,
    revenue INT
);

INSERT INTO sales VALUES
  ('Electronics', 'Phone', 1000),
  ('Electronics', 'Laptop', 2000),
  ('Electronics', 'Tablet', 1500),
  ('Clothing', 'Shirt', 500),
  ('Clothing', 'Jacket', 800);
```

```python
%sql
WITH ranked_sales AS (
  SELECT
    category,
    product,
    revenue,
    ROW_NUMBER() OVER (PARTITION BY category ORDER BY revenue DESC) AS rn
  FROM sales
)
SELECT
  category,
  product,
  revenue
FROM ranked_sales
WHERE rn = 1;
```

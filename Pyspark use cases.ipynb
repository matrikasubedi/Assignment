Do's and Don't

- API's -> from API's we take raw data - Bronze Layer -> cleaned layer(silver layer) ->Aggregration layer(gold layer) -> Visulation tool

Do's

1. Filtertion early (reduce data as fast as it can)

df = df.filter(col("event_date>="2023-04-31"))

2. select only required columns:
df = df.select("id","amount")

3. use proper partition

df.write .partitionBy("event_date").parquet("path")

Read
df = spark.read.parquet("path").filter("event_date=='2012-05-06'")

4. Use Broadcast join: broadcast for small table

5. Handle your bad data explicitly 

spark.read.option("mode","permissive mode")

- store corrupt data separately

6. Cache: We can use cache only when needed

7. Handle data skew


Don't

1. Don't use "collect" on large data sets
  - df.collect()

2. Don't use repatition
   - df.repartition(1000)

3. Don't use coalesce
    
4. don't ignore null handling because null can breaks aggregation silently

5. dont use udf's(user define function)- it makes optimization slow

Data skew

- when data is unevenly distributed across your partition

solution: re-partition ,salting

collect
- Brings all data from cluster to driver mode 
- use when we have small data
- df.limit(10).collect()

UDF's (user define function)

- def to_upper(X):\
    return X.upper()

Caching:
  - df.cache
  - where we store DataFrame in memory

Caching vs Persist

storage - for caching we have memory only, persist we have memory + disk

Transformation:
 Narrow : Each outpur partition depends only on input partition
 - no suffle happens

- df = spark.read.parquet("orders")
- df2 = df.filter(df.amount>100)
- df3 = df2.withColumn("amount_tax",df.amount*1.18)
- no data exchange between partitions
- eg.filter,select,withcolumn,drop
- Faster then wide

Wide: where data from multiple partition must be combined
 - cause suffle
 - data movement across nodes
 - df.groupBy("country)
 - data is resuffled across your cluster
 - eg: groupby, re-partition, join, orderBy
 - expensive and slow
